```python
import random

def word_guessing_game():

    words = ["python", "machine", "learning", "bridge"]

    secret_word = random.choice(words)

    guessed_letters = []
    wrong = 0

    print("Welcome to the Word Guessing Game")

    while wrong < 6:
        display = ""
        for letter in secret_word:
            if letter in guessed_letters:
                display = display + letter + " "
            else:
                display = display + "_ "
                print("Word :", display)

        if "_" not in display:
            print("Congratulations! You guessed the word.")
            break
        guess = input("Enter one letter: ").lower()
        if len(guess) != 1:
            print("Please enter only one letter.")
            continue
        if guess in guessed_letters:
            print("Letter already entered.")
            continue

        guessed_letters.append(guess)

        if guess in secret_word:
            print("Correct Guess!")
        else:
            wrong = wrong + 1
            print("Wrong Guess!")
            print("Remaining Attempts:", 6 - wrong)

    if wrong == 6:
        print("Game Over!")
        print("Correct Word is:", secret_word)

word_guessing_game()
```
